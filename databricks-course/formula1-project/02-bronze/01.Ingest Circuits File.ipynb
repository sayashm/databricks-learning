```python
source_file = f'{landing_folder_path}/circuits.csv'
table_name = f'{catalog_name}.{bronze_schema}.circuits'
```

## Step 1 - Read the CSV file using the dataframe reader API

```python
from pyspark.sql.types import StructType, StructField, StringType, DoubleType

circuits_schema = StructType([
    StructField('circuitID',    StringType()),
    StructField('url',          StringType()),
    StructField('circuitName',  StringType()),
    StructField('lat',          DoubleType()),
    StructField('long',         DoubleType()),
    StructField('locality',     StringType()),
    StructField('country',      StringType())

])
```

```python
circuits_df = (
    spark.read
    .format('csv')
    .option('header', 'true')       # first line to header
#   .option('inferSchema', 'true')  # imply format to each column
    .schema(circuits_schema)        #schema types
    .option('mode', 'FAILFAST')
    .load(source_file)
    )
```

```python
display(circuits_df)
# circuits_df.show()
```

## Step 2 - Add Metadata Columns
- Source File
- Ingestion Timestamp

```python
circuits_final_df = add_ingetion_metadata(circuits_df)
```

```python
display(circuits_final_df)
```

## Step 3 - Write to bronze delta table

```python
(
    circuits_final_df
        .write
        .format('delta')
        .mode('overwrite')
        .saveAsTable(table_name)
)
```
