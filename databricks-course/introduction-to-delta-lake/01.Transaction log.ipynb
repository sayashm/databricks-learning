# Understanding Delta Lake Transaction Log
Understand the cload storage directory structure for delta table

### 1. Create Catalog and Schema for the Demo
**Catalog:** Demo\
**Schema:** delta_lake

```sql
CREATE CATALOG IF NOT EXISTS demo
MANAGED LOCATION 'abfss://demo@databrickscourseextdl136.dfs.core.windows.net'
```

```sql
CREATE SCHEMA IF NOT EXISTS demo.delta_lake
MANAGED LOCATION 'abfss://demo@databrickscourseextdl136.dfs.core.windows.net/delta_lake'
```

### 1. Create a Delta Lake Table

```sql
CREATE TABLE IF NOT EXISTS demo.delta_lake.companies
(
    company_name STRING,
    founded_date Date,
    Country      STRING
)
USING DELTA;
```

```sql
DESC EXTENDED demo.delta_lake.companies;
```
